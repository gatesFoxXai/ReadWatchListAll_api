# 台積電 (2330) 盤中資料分析

這是一個 **Jupyter Notebook** 示範，使用本專案的真實資料 `2330.csv`（台積電盤中逐筆/分筆資料）。

Jupyter 的特色：
- 每個 cell 獨立執行，可反覆試錯
- 程式碼 + 文字說明混排
- 圖表直接顯示在 cell 內

```python
import pandas as pd
import numpy as np
import matplotlib

matplotlib.use("Agg")
import matplotlib.pyplot as plt

print(f"pandas {pd.__version__}")
print(f"numpy  {np.__version__}")
```

```python
import os

csv_path = "2330.csv"
if not os.path.exists(csv_path):
    raise FileNotFoundError(f"找不到資料檔：{os.path.abspath(csv_path)}")

df = pd.read_csv(csv_path)
df["日期時間"] = pd.to_datetime(df["日期時間"])
df = df.set_index("日期時間").sort_index()

print("形狀:", df.shape)
df.head()
```

```python
summary = {
    "tick 數": len(df),
    "起始時間": df.index.min(),
    "結束時間": df.index.max(),
    "DealPrice 最低": df["DealPrice"].min(),
    "DealPrice 最高": df["DealPrice"].max(),
    "DealPrice 平均": round(df["DealPrice"].mean(), 2),
    "總成交量 (外盤+內盤)": int(df["TotalOutVol"].sum() + df["TotalInVol"].sum()),
}
for k, v in summary.items():
    print(f"{k}: {v}")
```

```python
plt.figure(figsize=(10, 4))
plt.plot(df.index, df["DealPrice"], linewidth=0.8, color="#1f77b4")
plt.title("台積電 2330 DealPrice 走勢")
plt.xlabel("時間")
plt.ylabel("DealPrice")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## 小結

- **互動式**：每個 cell 可單獨執行、反覆調整，不用重跑整個腳本
- **即時圖表**：`plt.show()` 的結果直接顯示在 cell 內
- **程式碼 + 敘述混排**：適合做資料探索、報告、教學

你可以繼續編輯任何 cell 再執行，或新增 cell 做進一步分析（例如分時段統計、成交量加權均價 VWAP 等）。
